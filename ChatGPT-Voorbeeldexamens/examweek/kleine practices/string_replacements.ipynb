```python
import pandas as pd

# Voorbeeld DataFrame
df = pd.DataFrame({
    'Naam': [' Anna ', 'Bob Smith', 'Charlie-Smith', 'Diana'],
    'Stad': ['Amsterdam', 'Rotterdam', 'Antwerpen', 'Den Haag']
})

print("Origineel:")
print(df)

# 1️⃣ Whitespace verwijderen (links/rechts)
df['Naam_clean'] = df['Naam'].str.strip()

# 2️⃣ Specifieke substring vervangen
df['Naam_no_smith'] = df['Naam'].str.replace("Smith", "", regex=False)

# 3️⃣ Meerdere tekens vervangen met regex
df['Naam_regex'] = df['Naam'].str.replace(r"[- ]", "_", regex=True)
# "-" en spatie worden "_"  

# 4️⃣ Alles lowercase / uppercase / titlecase
df['Naam_lower'] = df['Naam'].str.lower()
df['Naam_upper'] = df['Naam'].str.upper()
df['Naam_title'] = df['Naam'].str.title()

# 5️⃣ Alleen cijfers/letters behouden (regex)
df['Naam_letters'] = df['Naam'].str.replace(r"[^A-Za-z]", "", regex=True)

# 6️⃣ Substring vervangen in Stad kolom
df['Stad_replace'] = df['Stad'].str.replace("dam", "town", regex=False)
```
