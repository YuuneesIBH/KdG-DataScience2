```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Anna", "Bob", "Charlie", "Diana", "Eva", "Frank", "Grace", "Henry"],
    "Age": [25, 30, 35, 28, 22, 40, 29, 33],
    "City": ["Amsterdam", "Rotterdam", "Utrecht", "Den Haag", "Gent", "Antwerpen", "Brugge", "Leuven"],
    "Gender": ["Female", "Male", "Male", "Female", "Female", "Male", "Female", "Male"],
    "Score": [88, 92, 79, 85, 91, 73, 95, 80]
})
print(df)
print(df.shape)
```

```python
print(df.loc[1:2, ['Name', 'Age']])

print(df.iloc[6, [1]])
```

```python
print(df.iloc[1:3, [0,1]])
```
