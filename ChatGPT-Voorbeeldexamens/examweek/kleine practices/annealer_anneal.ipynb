# Simulated Annealing Oefenopgave: Restaurant Menu Optimalisatie

## Achtergrond
Je bent eigenaar van een klein restaurant en wil een dagmenu samenstellen.  
Je hebt een lijst van mogelijke gerechten, elk met een **kostprijs** en een **populiariteitsscore** (hoeveel klanten het gerecht waarschijnlijk zullen bestellen).  
Doel is om een combinatie van gerechten te kiezen die **binnen je budget blijft**, maar **zo populair mogelijk** is. Budget --> 40 euro

## Gegevens
Er zijn 10 mogelijke gerechten:

````python
kost = np.array([5, 12, 8, 6, 10, 7, 9, 4, 11, 3])  # euro
score = np.array([50, 90, 70, 40, 85, 55, 65, 30, 95, 25])  # populariteitspunten
gerechten = ['Pasta', 'Steak', 'Salade', 'Soep', 'Pizza', 'Burger',
             'Vis', 'Fries', 'Sushi', 'Dessert']

class DagSchoteProbleem(Annealer):
    def energy(self):
        mask = np.array(self.state, dtype=bool)
        
        totale_kost = kost[mask].sum()
        totale_score = score[mask].sum()
        
        if totale_kost > 40:
            return 9000000
        
        #- niet vergeten!!!!!
        return -totale_score
    
    def move(self):
        i = np.random.randint(0, len(self.state))
        self.state[i] = 1 - self.state[i]
        
#starten met random state
initial_state = np.random.choice([0,1], size=10) #size = aantal items waaruit je keuze hebt
dagschotel_problem = DagSchoteProbleem(initial_state)
beste_oplossing, beste_score = dagschotel_problem.anneal()

print(f"Optimale rugzak: {beste_oplossing}")
print(f"Totaal kosten: {kost[np.array(beste_oplossing, dtype=bool)].sum()}kg")
print(f"Totale score: {score[np.array(beste_oplossing, dtype=bool)].sum()} punten")
````

````python
#optioneel
print("Optimale menu samenstelling:")
for gerecht in np.array(gerechten)[mask]:
    print(" -", gerecht)

print(f"Totaal kost: {kost[mask].sum()} euro")
print(f"Totale score: {score[mask].sum()} punten")
````
