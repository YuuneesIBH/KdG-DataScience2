```python
# Annealer oefening: Rastrigin functie minimaliseren
# Vereisten: pip install simanneal

import numpy as np
import math
from simanneal import Annealer

# Definieer het probleem
class RastriginProblem(Annealer):
    def move(self):
        # Kies willekeurig x of y en maak een kleine aanpassing
        i = np.random.randint(0, 2)  
        self.state[i] += np.random.normal(0, 0.1)
        # Zorg dat waarden binnen [-5.12, 5.12] blijven
        self.state[i] = np.clip(self.state[i], -5.12, 5.12)

    def energy(self):
        # Doelfunctie = Rastrigin (te minimaliseren)
        x, y = self.state
        return (20 + x**2 - 10*math.cos(2*math.pi*x) 
                   + y**2 - 10*math.cos(2*math.pi*y))

# Starttoestand (random punt in [-5.12, 5.12]^2)
init_state = np.random.uniform(-5.12, 5.12, size=2)

# Probleem initialiseren
rastrigin = RastriginProblem(init_state)

# Optioneel parameters instellen
rastrigin.Tmax = 25000.0   # starttemperatuur
rastrigin.Tmin = 2.5       # eindtemperatuur
rastrigin.steps = 50000    # aantal iteraties
rastrigin.updates = 100    # hoeveel statusupdates printen

# Run simulated annealing
best_state, best_energy = rastrigin.anneal()

print("Beste oplossing (x,y):", best_state)
print("Kostfunctie waarde:", best_energy)
```
