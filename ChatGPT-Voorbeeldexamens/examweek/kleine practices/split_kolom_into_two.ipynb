```python
df = pd.DataFrame({
    "Name": ["Alice Smith", "Bob Johnson", "Charlie Brown", "Diana White", "Eva Green"]
})

# 1. Simpele vervanging
df["Name"] = df["Name"].str.replace("Smith", "⭐", regex=False)

# 2. Alles lowercase maken en 'a' → '@'
df["Name"] = df["Name"].str.lower().str.replace("a", "@", regex=False)

# 3. Regex gebruiken: vervang alle spaties door '_'
df["Name"] = df["Name"].str.replace(r"\s+", "_", regex=True)

print(df)
```

```python
# Letters
# [A-Za-z]    → alle letters
# [A-Z]       → hoofdletters
# [a-z]       → kleine letters

# Cijfers
# [0-9]       → cijfers
# \d          → cijfers (digit)
# \D          → niet-cijfers

# Whitespace (spaties e.d.)
# \s          → whitespace (spatie, tab, newline)
# \S          → alles behalve whitespace
# " "         → gewone spatie

# Specifieke tekens
# \.          → punt
# \,          → komma
# \-          → minteken/streep
# _           → underscore
# @           → at-tekentje
# #           → hashtag
# \$          → dollar (moet escaped worden)
# %           → procent
# &           → ampersand
# \+          → plus
# \*          → sterretje
# \/          → slash
# \\          → backslash

# Symbolische klassen
# \w          → "word characters" (letters, cijfers, underscore)
# \W          → niet-woordtekens (bv. spaties, symbolen)
# \b          → woordgrens
# .           → gelijk welk teken (behalve newline)

# Andere handige regex dingen
# ^           → begin van string
# $           → einde van string
# ?           → optioneel
# +           → 1 of meer keer
# *           → 0 of meer keer
```
