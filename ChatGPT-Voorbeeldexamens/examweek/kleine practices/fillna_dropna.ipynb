```python
import pandas as pd

# Voorbeeld DataFrame
df = pd.DataFrame({
    'Naam': ['Anna', 'Bob', None, 'Diana'],
    'Leeftijd': [25, None, 35, 28],
    'Stad': ['Amsterdam', 'Rotterdam', 'Utrecht', None]
})

print("Origineel:")
print(df)

# 1️⃣ Alle NaN waarden in één kolom verwijderen
df_kolom_clean = df.dropna(subset=['Leeftijd'])

# 2️⃣ Alle NaN waarden in de hele dataset verwijderen (rijen met minstens 1 NaN weg)
df_clean = df.dropna()

# 2b️⃣ Kolommen met NaN verwijderen
df_kolommen_clean = df.dropna(axis=1)

# 3️⃣ NaN waarden vervangen door een specifieke waarde (hele dataset)
df_vervang_0 = df.fillna(0)                # Alle NaN → 0
df_vervang_str = df.fillna("Onbekend")     # Alle NaN → "Onbekend"

# 3b️⃣ Alleen in bepaalde kolom NaN vervangen
df_specifiek = df.copy()
df_specifiek['Leeftijd'] = df_specifiek['Leeftijd'].fillna(99)  # Leeftijd → 99
df_specifiek['Naam'] = df_specifiek['Naam'].fillna("Geen naam") # Naam → "Geen naam"

# 4️⃣ Slimme opvulling
df_mean = df.copy()
df_mean['Leeftijd'] = df_mean['Leeftijd'].fillna(df_mean['Leeftijd'].mean())

df_ffill = df.fillna(method='ffill')  # Forward fill (neem vorige waarde)
df_bfill = df.fillna(method='bfill')  # Backward fill (neem volgende waarde)
```
