## 📄 Examenvraag – Optimale Fietsconfiguratie (Simulated Annealing)

We willen de **beste fiets** samenstellen uit een reeks beschikbare onderdelen (A‐ en B‐onderdelen).  
- **Doel**: maximale kwaliteit, maar de totale prijs mag niet hoger zijn dan **€2500**.  
- **Oplossing** wordt voorgesteld als een **binaire array** waarin `1` betekent dat het onderdeel gekozen is.  
- **Prijsberekening**: prijs = som van de prijzen van de geselecteerde onderdelen.  
- **Kwaliteitsberekening**: kwaliteit = som van de kwaliteitscores van de geselecteerde onderdelen.  
- **Constraints**:
  - Er moet minstens één A‐onderdeel en één B‐onderdeel gekozen zijn.
  - Als prijs > €2500 → **strafwaarde** geven zodat de oplossing altijd slechter is.

````python
#imports
import numpy as np
from simanneal import Annealer
````

````python
A_kw = np.array([9, 8, 9, 9, 8, 9, 9, 9, 8, 9])
B_kw = np.array([7, 6, 7, 8, 6, 6, 8, 8, 7, 6])

A_prijs = np.array([1300, 250, 350, 200, 300, 320, 150, 250, 450, 160])
B_prijs = np.array([200, 100, 300, 150, 250, 220,  50, 180, 350,  20])

class BikeProblem(Annealer):
    def move(self):
        i = np.random.randint(0, len(self.state))
        self.state[i] = 1 - self.state[i] #dit hier is de toggle
        
    def energy(self):
        A_staat = self.state[:len(A_kw)]
        B_staat = self.state[len(A_kw):]
        
        #totaalprijs - 2500 max
        total = A_prijs[A_staat == 1].sum() + B_prijs[B_staat == 1].sum()
        #quality
        total_quality = A_kw[A_staat == 1].sum() + B_kw[B_staat == 1].sum()
        
        if total > 2500 or A_staat.sum() == 0 or B_staat.sum() == 0:
            return 1e9
        
        return -total_quality
````

````python
initial_state = np.random.choice([0,1], size=10)
bikeproblem = BikeProblem(initial_state)
quality = bikeproblem.anneal()
````

````python
print(f'De array voor de optimale fiets met een prijs van maximum 2500 euro:\n' f'{quality[0]}')
````

````python
best_state, best_energy = bikeproblem.anneal()

A_state = best_state[:len(A_kw)]
B_state = best_state[len(A_kw):]

total_price = A_prijs[A_state == 1].sum() + B_prijs[B_state == 1].sum()

print(f"De prijs van de optimale fiets: {total_price}")
````
