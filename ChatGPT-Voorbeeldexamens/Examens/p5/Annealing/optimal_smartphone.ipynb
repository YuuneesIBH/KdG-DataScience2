## 📄 Examenvraag – Optimale Smartphone-configuratie (Simulated Annealing)

We willen de **beste smartphone** samenstellen uit een reeks beschikbare onderdelen (A- en B-onderdelen).  
- **Doel**: maximale kwaliteit, maar de totale prijs mag niet hoger zijn dan **€1500**.  
- **Oplossing** wordt voorgesteld als een **binaire array** waarin `1` betekent dat het onderdeel gekozen is.  
- **Prijsberekening**: prijs = som van de prijzen van de geselecteerde onderdelen.  
- **Kwaliteitsberekening**: kwaliteit = som van de kwaliteitscores van de geselecteerde onderdelen.  
- **Constraints**:
  - Er moet minstens één A-onderdeel en één B-onderdeel gekozen zijn.
  - Als prijs > €1500 → **strafwaarde** geven zodat de oplossing altijd slechter is.

Je krijgt de volgende data:

```python
A_kw = np.array([8, 9, 7, 6, 9])

````python
#imports
import numpy as np
from simanneal import Annealer
````

````python
A_kw = np.array([8, 9, 7, 6, 9])
B_kw = np.array([7, 8, 6, 9, 7])

A_pr = np.array([400, 500, 350, 300, 450])
B_pr = np.array([350, 420, 280, 500, 320])

class SmartphoneProblem(Annealer):
    def move(self):
        i = np.random.randint(0, len(self.state))
        self.state[i] = 1 - self.state[i] #dit hier is de toggle
        
    def energy(self):
        #kwaliteit van A en B toekennen aan de staat
        A_staat = self.state[:len(A_kw)]
        B_staat = self.state[len(A_kw):]
        
        #totaalprijs berekenen 
        total = A_pr[A_staat == 1].sum() + B_pr[B_staat == 1].sum()
        #kwaliteit van de smartphone berekenen
        quality = A_kw[A_staat == 1].sum() + B_kw[B_staat == 1].sum()
        
        #error geven als de waarde boven 1500 is
        if total > 1500 or A_staat.sum() == 0 or B_staat.sum() == 0:
            return 1e9
        
        return -quality
````

````python
#een random genereren om er gewoon van te vertrekken 
initial_state = np.random.choice([0, 1], size=10)
smartphoneproblem = SmartphoneProblem(initial_state)
quality = smartphoneproblem.anneal()
````

````python

print(f'De array voor de optimale smartphone met een prijs van maximum 1500 euro:\n' f'{quality[0]}')
````
