# The difference between loc and iloc

```python
import pandas as pd
rides = {"Ride Name" : ["Titan", "New Texas Giant", "Batman", "Mr Freeze", "Shockwave"],
         "Inversions" : [0, 0, 5, 1, 2],
         "Top Speed (mph)" : [85, 65, 50, 70, 60],
         "Height (ft)" : [245, 153, 105, 218, 116]}
df = pd.DataFrame(data=rides)
```

```python
print(df)
```

```python
print(df.loc[2])
```

```python
print(df.iloc[-1])
```
