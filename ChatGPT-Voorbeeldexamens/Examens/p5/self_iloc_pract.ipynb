### loc vs iloc self practice

```python
import seaborn as sns
import pandas as pd
```

```python
df = sns.load_dataset('iris')
print(df.head(30))
```

```python
#de eerste vier rijen afdrukken
print(df.loc[:3])
```

```python
#stel ik wil de derde rij hebben
print(df.loc[2])
```

```python
#stel ik wil de derde rij tot de vijfde hebben
print(df.loc[2:4])
```

```python
#ik wil nu van de tweede rij tot de vijfde de petal lenght hebben
print("Rij 2, kolom 'petal_length':", df.loc[2:4, 'petal_length'])
```

```python
#ik wil nu van rij 29 de petal_width hebben
print(df.loc[29, 'petal_width'])
```

```python
#nu wil ik de petal width selecteren van rij 6 en 26!
print(df.loc[[6, 29], "petal_width"])
```

```python
#nu beginnen met iloc --> selecteren van sepal length en widht van 14 en 16
df.iloc[[14, 16], [0,3]]
```

```python
print(df.loc[14, "sepal_length"])
```

# iloc is vanaf index 0 tellen!!!!!!

```python
print(df.iloc[14, 0])
```

```python
print(df.iloc[10:21, [0, 4]])
```

```python
print(df.loc[10:21, ["sepal_length", "species"]])
```
