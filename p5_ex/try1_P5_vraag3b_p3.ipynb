```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#import van de csv
df = pd.read_csv("inzamelingenALL.csv", sep=";", decimal=",")
df
```

```python
df = df[df["Jaar"].between(2021, 2024)]
```

```python
#filteren om enkel bepaalde jaren te selecteren en ervoor zorgen dat er geen 0 gewichten doorkomen
df_per_jaar = df.groupby("Jaar")["kostprijs"].sum()
```

```python
df_per_jaar.plot(marker="o")
plt.title("Kostprijs per jaar")
plt.xlabel("Jaar")
plt.ylabel("Kostprijs in EUR")
plt.grid(True)
plt.show()
```
