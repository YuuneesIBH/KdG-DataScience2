```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter
```

```python
#import van de csv file
df_afval = pd.read_csv("inzamelingenALL.csv", sep=";", decimal=",")
```

```python
gewichten = df_afval["Gewicht (kg)"]
gewichten = gewichten[gewichten > 0]
```

```python
#histogram tekenen 
plt.hist(gewichten, bins=20, alpha=0.3, density=True)

plt.ylabel("Percentage")
plt.gca().yaxis.set_major_formatter(PercentFormatter(1))  # 1 = 100%
plt.title("Histogram van Gewichten")
plt.show()
```
